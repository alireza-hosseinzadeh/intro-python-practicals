## 5. Variables and assignment

A **variable** is a name that refers to a value or object in memory.

In Python, we use `=` for assignment.

For example:

```python

```

This means: assign the value `30` to the variable name `age`.

It does **not** mean mathematical equality.

````python

# Create some variables

name = "Ali"
age = 30
height = 1.75

# Print the variables
print(name)
print(age)
print(height)
````

### Variable naming conventions

Good variable names should be clear and descriptive.

Recommended style in Python:

```python
student_name = "Ali"
exam_score = 72
average_income = 31500
```

Avoid unclear names such as:

```python
x = "Ali"
a = 72
thing = 31500
```

Also avoid spaces in variable names. Use underscores instead.

````python

# Good variable names

student_name = "Sara"
exam_score = 85
course_name = "Introductory Python"

print(student_name)
print(exam_score)
print(course_name)
````

## 7. Strings

A **string** is text. In Python, strings can be written using single or double quotation marks.

Examples:

```python
"Manchester"
'Python'
```

You can combine strings using `+`. This is called concatenation.

````python

# String examples

first_name = "Ada"
last_name = "Lovelace"

full_name = first_name + " " + last_name

print(full_name)
print(full_name.upper())      # convert to uppercase
print(full_name.lower())      # convert to lowercase
print(len(full_name))         # count the number of characters
````

### Formatted strings

An f-string lets us insert variable values directly into text.

````python

# f-string example

student = "Sara"
score = 82

print(f"{student} scored {score} in the test.")
````

## 8. Lists

A **list** is an ordered and changeable collection of values.

Lists are written using square brackets `[]`.

````python

# List example

scores = [65, 72, 81, 90, 58]

print(scores)
print(scores[0])      # first item; indexing starts at 0
print(scores[1])      # second item
print(scores[-1])     # last item
print(len(scores))    # number of items in the list
````

### List methods

Lists have useful methods such as:

- `.append()` to add an item;
- `.remove()` to remove an item;
- `.sort()` to sort the list.

````python

# Working with list methods

scores = [65, 72, 81, 90, 58]

scores.append(77)     # add a new score to the end
print(scores)

scores.remove(58)     # remove the value 58
print(scores)

scores.sort()         # sort from smallest to largest
print(scores)
````

## 9. Tuples

A **tuple** is an ordered collection, similar to a list, but it cannot be changed after creation.

Tuples are written using parentheses `()`.

They are useful for fixed values, such as coordinates.

````python

# Tuple example

coordinates = (53.4808, -2.2426)  # approximate coordinates for Manchester

print(coordinates)
print(coordinates[0])  # latitude
print(coordinates[1])  # longitude
````

## 10. Sets

A **set** is an unordered collection of unique values.

Sets are written using curly braces `{}`.

They are useful when duplicates should be removed automatically.

````python

# Set example

programming_languages = {"Python", "R", "SQL", "Python"}

# The repeated "Python" appears only once because sets contain unique values.
print(programming_languages)

programming_languages.add("JavaScript")
print(programming_languages)

programming_languages.remove("R")
print(programming_languages)
````

### Set operations

Sets are useful for comparing groups.

````python

# Set operations

python_students = {"Ali", "Sara", "John", "Mei"}
r_students = {"Sara", "Mei", "David"}

print(python_students.union(r_students))          # students in either group
print(python_students.intersection(r_students))   # students in both groups
print(python_students.difference(r_students))     # in Python group but not R group
````

## 11. Dictionaries

A **dictionary** stores data as key-value pairs.

Dictionaries are written using curly braces `{}` with the form:

```python
{"key": "value"}
```

They are useful when each value has a label.

````python

# Dictionary example

student = {
    "name": "Ali",
    "age": 30,
    "city": "Manchester"
}

print(student)
print(student["name"])  # access a value using its key
print(student["city"])
````

### Updating dictionaries

You can add or change a dictionary value using square brackets.

````python

# Updating a dictionary

student = {
    "name": "Ali",
    "age": 30
}

student["city"] = "Manchester"  # add a new key-value pair
student["age"] = 31              # update an existing value

print(student)
print(student.keys())             # show all keys
print(student.values())           # show all values
````

## 12. Conditional statements

A conditional statement lets Python make decisions.

The basic structure is:

```python
if condition:
    code to run if condition is true
else:
    code to run otherwise
```

Indentation is essential in Python. The indented lines belong to the `if` or `else` block.

````python

# Conditional statement example

score = 68

if score >= 50:
    print("Pass")
else:
    print("Fail")
````

````python

# Multiple conditions using if, elif, and else

score = 75

if score >= 70:
    print("Distinction")
elif score >= 60:
    print("Merit")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
````

## 13. Loops

A loop repeats code.

A `for` loop is commonly used to go through items in a collection.

````python

# Loop through a list

scores = [65, 72, 81, 90, 58]

for score in scores:
    print(score)
````

````python

# Use a loop to calculate adjusted scores

scores = [65, 72, 81, 90, 58]

for score in scores:
    adjusted_score = score + 5  # add 5 marks to each score
    print(adjusted_score)
````

## 14. Simple data-analysis example

We can combine variables, lists, and loops to calculate summary statistics.

Here we calculate the mean manually.

````python

# A small dataset of exam scores

scores = [65, 72, 81, 90, 58]

# sum(scores) adds all values in the list.
# len(scores) counts how many values are in the list.
mean_score = sum(scores) / len(scores)

print(f"The mean score is {mean_score}.")
````

### Using a package

Python has many packages. A package is a collection of extra tools that extend Python.

For numerical work, one common package is `numpy`.

We usually import packages at the top of a notebook or script.

````python

# Import numpy and give it the standard short name np

import numpy as np

scores = np.array([65, 72, 81, 90, 58])

print(np.mean(scores))   # mean
print(np.min(scores))    # minimum
print(np.max(scores))    # maximum
print(np.std(scores))    # standard deviation
````

## 15. Common errors

Errors are normal when programming. The important skill is learning how to read them.

Common categories:

| Error type | Meaning | Example |
|---|---|---|
| Syntax error | Python cannot understand the code structure | missing parenthesis |
| Runtime error | Code starts running but fails during execution | dividing by zero |
| Logical error | Code runs but gives the wrong result | wrong formula |

````python

# Logical error example
# Suppose we want the average of two numbers: 10 and 20.
# The correct formula is (10 + 20) / 2.
# The following formula runs, but it gives the wrong answer because of operator precedence.

wrong_average = 10 + 20 / 2
correct_average = (10 + 20) / 2

print(wrong_average)
print(correct_average)
````

## 16. Debugging with print statements

A simple way to debug code is to print intermediate values.

This is called tracing because we trace what happens step by step.

````python

# Debugging example using print statements

price = 100
discount_rate = 0.20

print("Original price:", price)
print("Discount rate:", discount_rate)

discount = price * discount_rate
print("Discount:", discount)

final_price = price - discount
print("Final price:", final_price)
````

## 17. Writing clear comments

Comments should explain why something is done, not merely repeat what the code already says.

Weak comment:

```python
x = x + 1  # add 1 to x
```

Better comment:

```python
x = x + 1  # move to the next time period
```

Use comments to make your code easier to review, especially when working with others or using LLM-assisted coding.

## 18. Using LLMs carefully for coding

LLMs can help generate, explain, and debug code. However, the user must still check that the code does what was intended.

A good prompt should include:

- the task goal;

Example prompt:

After receiving code from an LLM, check:

- whether the code runs;
- whether the output is plausible;
