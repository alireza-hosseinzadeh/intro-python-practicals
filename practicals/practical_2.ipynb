## 2. Reminder: objects, names, and assignment

In Python, the `=` sign assigns a **name** to an object. The variable name does not store the object directly; it refers to where the object is stored in memory.

Several names can refer to the same object.


```python
# A simple assignment
x = [1, 2, 3]

# y now refers to the same list object as x
y = x

print("x:", x)
print("y:", y)
```

Because lists are **mutable**, changing the object through one name is visible through the other name.


```python
# Add a new value to the list using y
y.append(4)

# x has also changed because x and y refer to the same list object
print("x:", x)
print("y:", y)
```

To create an independent copy of a list, use `.copy()` or slicing.


```python
a = [1, 2, 3]
b = a.copy()   # independent copy

b.append(4)

print("a:", a)
print("b:", b)
```

## 3. Lists

A **list** is an ordered, mutable collection. Lists are useful when we want to store several values and possibly change them later.


```python
# A list of course names
courses = ["Python", "R", "SQL"]

print(courses)
print(type(courses))
```

### 3.1 Indexing and slicing

Python indexing starts at `0`, so the first element is at index `0`. Negative indexes count from the end.


```python
courses = ["Python", "R", "SQL", "Excel"]

print("First course:", courses[0])
print("Second course:", courses[1])
print("Last course:", courses[-1])
print("First two courses:", courses[:2])
```

### 3.2 Updating list values

Lists are mutable, so we can replace an element at a given index.


```python
courses = ["Python", "R", "SQL"]

# Replace the second element
courses[1] = "Stata"

print(courses)
```

### 3.3 Adding values to a list

Common list methods:

| Method | Purpose |
|---|---|
| `.append(x)` | Add `x` to the end of the list |
| `.insert(i, x)` | Add `x` at index `i` |
| `.extend(other_list)` | Add all values from another list |


```python
courses = ["Python", "R", "SQL"]

# Add one item to the end
courses.append("Excel")
print(courses)

# Insert one item at a specific index
courses.insert(1, "Statistics")
print(courses)

# Add multiple items from another list
courses.extend(["NumPy", "Pandas"])
print(courses)
```

#### Important correction: call methods on the list object

Use `courses.insert(...)`, not `list.insert(...)`. The name `list` refers to the built-in list type, not to your specific list.


```python
courses = ["Python", "R", "SQL"]

# Correct: call insert on the object we want to modify
courses.insert(2, "Physical Exercise")

print(courses)
```

### 3.4 Removing values from a list

Common methods:

| Method | Purpose |
|---|---|
| `.remove(x)` | Remove the first occurrence of value `x` |
| `.pop()` | Remove and return the last item |
| `.pop(i)` | Remove and return the item at index `i` |
| `.clear()` | Remove all items |


```python
courses = ["Python", "R", "SQL", "Excel"]

# Remove by value
courses.remove("R")
print(courses)

# Remove by position and keep the removed value
removed_course = courses.pop(1)
print("Removed:", removed_course)
print("Remaining:", courses)
```

### 3.5 Sorting lists

There are two common ways to sort:

| Operation | Modifies original list? | Returns |
|---|---:|---|
| `list_name.sort()` | Yes | `None` |
| `sorted(list_name)` | No | A new sorted list |

Use `.sort()` when you want to change the original list. Use `sorted()` when you want to keep the original list unchanged.


```python
numbers = [4, 1, 3, 2]

# sorted() returns a new list and leaves the original unchanged
new_numbers = sorted(numbers)

print("Original:", numbers)
print("Sorted copy:", new_numbers)
```

```python
numbers = [4, 1, 3, 2]

# .sort() modifies the original list in place
numbers.sort()

print(numbers)
```

## 4. Tuples

A **tuple** is an ordered collection, similar to a list, but it is **immutable**. Once created, its elements cannot be changed.

Tuples are useful for fixed pieces of information, such as coordinates.


```python
coordinates = (53.48, -2.24)

print(coordinates)
print(type(coordinates))
print("Latitude:", coordinates[0])
print("Longitude:", coordinates[1])
```

Trying to change a tuple element will produce an error because tuples are immutable.


```python
coordinates = (53.48, -2.24)

# Uncomment the next line to see the error
# coordinates[0] = 50.00
```

### Tuple unpacking

Tuple unpacking lets us assign tuple values to separate variables.


```python
coordinates = (53.48, -2.24)

latitude, longitude = coordinates

print(latitude)
print(longitude)
```

## 5. Sets

A **set** is an unordered collection of unique values. Sets are useful when we want to remove duplicates or compare groups.


```python
courses = {"Python", "R", "SQL", "Python"}

# Duplicate values are automatically removed
print(courses)
print(type(courses))
```

### 5.1 Set methods

Common set methods:

| Method | Purpose |
|---|---|
| `.add(x)` | Add one value |
| `.remove(x)` | Remove a value; error if missing |
| `.discard(x)` | Remove a value; no error if missing |
| `.union(other)` | Values in either set |
| `.intersection(other)` | Values in both sets |
| `.difference(other)` | Values in the first set but not the second |
| `.symmetric_difference(other)` | Values in either set but not both |


```python
python_students = {"Ali", "Sara", "Tom", "Mina"}
r_students = {"Sara", "John", "Mina", "David"}

print("Union:", python_students.union(r_students))
print("Intersection:", python_students.intersection(r_students))
print("Python but not R:", python_students.difference(r_students))
print("One course only:", python_students.symmetric_difference(r_students))
```

### Set complement in Python

Python does not have a separate `.complement()` method. A complement is defined relative to a **universal set**.

For example, if `all_students` is the universal set, then the complement of `python_students` is:


```python
all_students = {"Ali", "Sara", "Tom", "Mina", "John", "David", "Nina"}
python_students = {"Ali", "Sara", "Tom", "Mina"}

not_python_students = all_students.difference(python_students)

print(not_python_students)
```

## 6. Dictionaries

A **dictionary** stores data as key-value pairs. Dictionaries are useful when each value has a meaningful label.


```python
student = {
    "name": "Ali",
    "age": 30,
    "city": "Manchester"
}

print(student)
print(type(student))
```

### 6.1 Accessing dictionary values

Use square brackets with the key name to access a value.


```python
student = {"name": "Ali", "age": 30, "city": "Manchester"}

print(student["name"])
print(student["age"])
```

The `.get()` method is safer when the key might not exist.


```python
student = {"name": "Ali", "age": 30}

print(student.get("name"))
print(student.get("city"))              # returns None if missing
print(student.get("city", "Unknown"))  # custom default value
```

### 6.2 Adding and updating dictionary values

We can add a new key-value pair using square brackets. This is often the clearest method.


```python
student = {"name": "Ali", "age": 30}

# Add a new key-value pair
student["city"] = "Manchester"

print(student)
```

The `.update()` method is useful when we want to add or update multiple key-value pairs at once.


```python
student = {"name": "Ali", "age": 30}

student.update({
    "city": "Manchester",
    "course": "Python"
})

print(student)
```

### 6.3 Dictionary methods

Common methods:

| Method | Purpose |
|---|---|
| `.keys()` | Return the keys |
| `.values()` | Return the values |
| `.items()` | Return key-value pairs |
| `.update()` | Add or update key-value pairs |
| `.pop(key)` | Remove a key and return its value |


```python
student = {"name": "Ali", "age": 30, "city": "Manchester"}

print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
```

## 7. Loops over data structures

Loops allow us to repeat an action for every element in a collection.


```python
courses = ["Python", "R", "SQL"]

for course in courses:
    print("I am learning", course)
```

### Looping over dictionaries

Use `.items()` when you need both the key and the value.


```python
student = {"name": "Ali", "age": 30, "city": "Manchester"}

for key, value in student.items():
    print(key, "->", value)
```

## 8. Mini task: student records

In this task, we combine lists and dictionaries. This is a common pattern in simple data work.


```python
# A list of dictionaries: each dictionary represents one student
students = [
    {"name": "Ali", "score": 78},
    {"name": "Sara", "score": 85},
    {"name": "Tom", "score": 62},
    {"name": "Mina", "score": 91},
]

# Print each student's name and score
for student in students:
    print(student["name"], "scored", student["score"])
```

### Calculate the average score


```python
students = [
    {"name": "Ali", "score": 78},
    {"name": "Sara", "score": 85},
    {"name": "Tom", "score": 62},
    {"name": "Mina", "score": 91},
]

# Extract the scores into a list
scores = []

for student in students:
    scores.append(student["score"])

# Calculate the average
average_score = sum(scores) / len(scores)

print("Average score:", average_score)
```
