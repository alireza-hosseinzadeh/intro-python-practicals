## 1. Importing the required libraries

Python libraries are collections of pre-written code.

In this practical:

| Library | Main use |
|---|---|
| `numpy` | Numerical calculations and arrays |
| `pandas` | Working with tabular data |
| `matplotlib.pyplot` | Basic plotting |
| `seaborn` | Statistical visualisation |

The usual convention is to import these libraries using short aliases:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

These aliases are not compulsory, but they are standard in data analysis.

````python
# Import NumPy for numerical operations.
import numpy as np

# Import pandas for working with tables/dataframes.
import pandas as pd

# Import Matplotlib for plotting.
import matplotlib.pyplot as plt

# Import Seaborn for statistical visualisation.
import seaborn as sns

# Import a built-in dataset from scikit-learn.
from sklearn.datasets import load_iris

# This command makes plots appear inside the notebook.
%matplotlib inline
````

## 2. Loading the dataset

The Iris dataset contains measurements for three species of iris flowers:

- `setosa`
- `versicolor`
- `virginica`

For each flower, the dataset records:

- sepal length
- sepal width
- petal length
- petal width

The measurements are in centimetres.

````python
# Load the Iris dataset.
iris = load_iris()

# The object returned by load_iris() is similar to a dictionary.
# We can inspect its keys to see what information it contains.
iris.keys()
````

The most important parts of this object are:

| Key | Meaning |
|---|---|
| `data` | The numerical measurements |
| `target` | The species label as numbers: 0, 1, 2 |
| `target_names` | The actual species names |
| `feature_names` | The names of the measurement variables |
| `DESCR` | A longer description of the dataset |

````python
# Print the feature names.
print(iris.feature_names)

# Print the target names.
print(iris.target_names)
````

## 3. Creating a pandas DataFrame

A `DataFrame` is the main pandas object for tabular data.

It is similar to:

- an Excel spreadsheet,
- an R data frame,
- or a table in a database.

Rows usually represent observations.  
Columns usually represent variables.

````python
# Convert the numerical data into a pandas DataFrame.
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Add the target column.
# This column contains numbers: 0, 1, and 2.
df["species_code"] = iris.target

# Convert the numeric species code into the actual species name.
# iris.target_names is an array: ['setosa', 'versicolor', 'virginica'].
df["species"] = df["species_code"].map(
    {0: "setosa", 1: "versicolor", 2: "virginica"}
)

# Display the first five rows of the DataFrame.
df.head()
````

## 4. First inspection of the dataset

Before analysing a dataset, always inspect it.

Common first checks include:

- How many rows and columns are there?
- What are the column names?
- What data types do the columns have?
- Are there any missing values?
- What are the first few rows?

````python
# Show the number of rows and columns.
# The output is a tuple: (number_of_rows, number_of_columns).
df.shape
````

````python
# Show the column names.
df.columns
````

````python
# Show the data types and non-missing counts.
df.info()
````

````python
# Count missing values in each column.
# A value of 0 means there are no missing values in that column.
df.isna().sum()
````

````python
# Display summary statistics for the numerical columns.
df.describe()
````

## 5. Renaming columns

The original column names contain spaces and units, for example:

```python
'sepal length (cm)'
```

For coding, it is often easier to use shorter column names without spaces.

This is not compulsory, but it makes the code cleaner.

````python
# Rename columns to shorter names.
df = df.rename(columns={
    "sepal length (cm)": "sepal_length",
    "sepal width (cm)": "sepal_width",
    "petal length (cm)": "petal_length",
    "petal width (cm)": "petal_width"
})

# Display the updated column names.
df.head()
````

## 6. Selecting columns

In pandas, a single column can be selected using square brackets.

For example:

```python
df["sepal_length"]
```

Multiple columns can be selected using a list of column names:

```python
df[["sepal_length", "species"]]
```

````python
# Select one column.
df["sepal_length"].head()
````

````python
# Select multiple columns.
df[["sepal_length", "petal_length", "species"]].head()
````

## 7. Filtering rows

Filtering means selecting only the rows that satisfy a condition.

For example, we may want only the flowers where:

```python
species == "setosa"
```

The condition produces a Boolean Series containing `True` and `False`.
Pandas keeps the rows where the condition is `True`.

````python
# Create a Boolean condition.
condition = df["species"] == "setosa"

# Show the first few values of the condition.
condition.head()
````

````python
# Use the condition to filter the DataFrame.
setosa_df = df[condition]

# Display the first five rows of the filtered data.
setosa_df.head()
````

````python
# Filter rows where petal_length is greater than 5.
long_petals = df[df["petal_length"] > 5]

# Display the first few matching rows.
long_petals.head()
````

## 8. Filtering with multiple conditions

When combining conditions in pandas:

| Operator | Meaning |
|---|---|
| `&` | and |
| `|` | or |
| `~` | not |

Each condition should be placed inside parentheses.

````python
# Select flowers that are virginica AND have petal length greater than 5.5.
large_virginica = df[
    (df["species"] == "virginica") & 
    (df["petal_length"] > 5.5)
]

large_virginica.head()
````

````python
# Select flowers that are either setosa OR have sepal length greater than 7.
setosa_or_large = df[
    (df["species"] == "setosa") | 
    (df["sepal_length"] > 7)
]

setosa_or_large.head()
````

## 9. Sorting data

Sorting allows us to order rows by the values of one or more columns.

For example, we can sort flowers from smallest to largest petal length.

````python
# Sort by petal_length in ascending order.
df.sort_values("petal_length").head()
````

````python
# Sort by petal_length in descending order.
df.sort_values("petal_length", ascending=False).head()
````

## 10. Creating new columns

New columns are often created from existing columns.

Here, we create:

- `sepal_area = sepal_length * sepal_width`
- `petal_area = petal_length * petal_width`

These are not true biological area measurements, but they are useful derived variables for practice.

````python
# Create a new column for approximate sepal area.
df["sepal_area"] = df["sepal_length"] * df["sepal_width"]

# Create a new column for approximate petal area.
df["petal_area"] = df["petal_length"] * df["petal_width"]

# Display the first few rows.
df.head()
````

## 11. Grouping and summarising data

Grouping is one of the most important data analysis operations.

The usual logic is:

1. Split the data into groups.
2. Apply a calculation to each group.
3. Combine the results into a summary table.

For example, we can calculate the average petal length for each species.

````python
# Calculate the mean of each numerical variable by species.
df.groupby("species").mean(numeric_only=True)
````

````python
# Calculate selected summary statistics for petal_length by species.
df.groupby("species")["petal_length"].agg(["count", "mean", "std", "min", "max"])
````

## 12. Using NumPy for numerical summaries

Pandas is built on top of NumPy.

A pandas column can be treated like a numerical array.

NumPy is useful for:

- means,
- medians,
- standard deviations,
- minimums and maximums,
- mathematical transformations.

````python
# Convert one pandas column to a NumPy array.
petal_lengths = df["petal_length"].to_numpy()

# Display the first ten values.
petal_lengths[:10]
````

````python
# Calculate numerical summaries using NumPy.
print("Mean:", np.mean(petal_lengths))
print("Median:", np.median(petal_lengths))
print("Standard deviation:", np.std(petal_lengths))
print("Minimum:", np.min(petal_lengths))
print("Maximum:", np.max(petal_lengths))
````

### Standard deviation note

By default, `np.std()` calculates the population standard deviation, using denominator `n`.

Pandas `Series.std()` calculates the sample standard deviation, using denominator `n - 1`.

That is why the two values may be slightly different.

````python
# NumPy population standard deviation.
print("NumPy population SD:", np.std(df["petal_length"]))

# NumPy sample standard deviation.
print("NumPy sample SD:", np.std(df["petal_length"], ddof=1))

# Pandas sample standard deviation.
print("Pandas sample SD:", df["petal_length"].std())
````

## 13. Basic plotting with Matplotlib

Matplotlib is a general-purpose plotting library.

The typical structure is:

```python
plt.figure()
plt.plot_or_chart_function()
plt.xlabel()
plt.ylabel()
plt.title()
plt.show()
```

We start with a histogram of petal length.

````python
# Create a histogram of petal length.
plt.figure(figsize=(8, 5))

# A histogram shows the distribution of a numerical variable.
plt.hist(df["petal_length"], bins=20)

# Add labels and a title.
plt.xlabel("Petal length")
plt.ylabel("Frequency")
plt.title("Distribution of Petal Length")

# Display the plot.
plt.show()
````

## 14. Scatter plots with Matplotlib

A scatter plot shows the relationship between two numerical variables.

Here, we compare petal length and petal width.

````python
# Create a scatter plot.
plt.figure(figsize=(8, 5))

# Each point represents one flower.
plt.scatter(df["petal_length"], df["petal_width"])

# Add labels and a title.
plt.xlabel("Petal length")
plt.ylabel("Petal width")
plt.title("Petal Length vs Petal Width")

# Display the plot.
plt.show()
````

## 15. Statistical visualisation with Seaborn

Seaborn is built on top of Matplotlib.

It is particularly useful for statistical plots because it works well with pandas DataFrames.

A common Seaborn pattern is:

```python
sns.plot_function(data=df, x="column1", y="column2", hue="grouping_column")
```

The `hue` argument colours the data by group.

````python
# Create a scatter plot using Seaborn.
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Petal Length vs Petal Width by Species")
plt.show()
````

## 16. Box plots

A box plot summarises the distribution of a numerical variable across groups.

It shows:

- the median,
- the interquartile range,
- and potential outliers.

Here, we compare petal length across species.

````python
# Create a box plot of petal length by species.
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)

plt.title("Petal Length by Species")
plt.show()
````

## 17. Violin plots

A violin plot is similar to a box plot, but it also shows the shape of the distribution.

Wider parts of the violin indicate where more observations are concentrated.

````python
# Create a violin plot of petal width by species.
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="species",
    y="petal_width"
)

plt.title("Petal Width Distribution by Species")
plt.show()
````

## 18. Pair plots

A pair plot shows relationships between several numerical variables at once.

This is useful for exploratory data analysis.

Because it creates many plots, it is better suited to datasets with a manageable number of columns.

````python
# Select the main numerical columns and the species column.
plot_df = df[[
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]]

# Create a pair plot.
sns.pairplot(plot_df, hue="species")

plt.show()
````

## 19. Correlation

Correlation measures the strength and direction of a linear relationship between numerical variables.

Values are between `-1` and `1`.

| Correlation | Interpretation |
|---|---|
| close to `1` | strong positive linear relationship |
| close to `0` | weak or no linear relationship |
| close to `-1` | strong negative linear relationship |

Correlation does not prove causation.

````python
# Calculate the correlation matrix for numerical variables.
correlation_matrix = df[[
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "sepal_area",
    "petal_area"
]].corr()

correlation_matrix
````

````python
# Visualise the correlation matrix using a heatmap.
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()
````

## 20. A simple analysis question

Suppose we want to answer:

> Which iris species tends to have the largest petals?

To answer this, we can compare average petal length and average petal area by species.

````python
# Compare average petal length and petal area by species.
species_summary = df.groupby("species")[["petal_length", "petal_width", "petal_area"]].mean()

species_summary
````

````python
# Plot average petal area by species.
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="species",
    y="petal_area",
    errorbar=None
)

plt.title("Average Petal Area by Species")
plt.xlabel("Species")
plt.ylabel("Average petal area")

plt.show()
````
