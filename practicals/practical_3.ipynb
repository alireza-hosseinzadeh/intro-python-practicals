## 3. Indentation

Python uses indentation to define code blocks. The standard convention is **4 spaces** for each indentation level.

A block starts after a line ending with a colon `:`.


````python
# The indented line belongs to the if-block.
x = 10

if x > 5:
    print("x is greater than 5")

# This line is outside the if-block.
print("This line always runs")
````

Incorrect indentation causes an `IndentationError`. For example, this would fail:

```python
if x > 5:
print("x is greater than 5")
```

The `print()` statement must be indented.


## 4. Conditional statements

Conditional statements allow a program to make decisions.

Main keywords:

- `if`: first condition;
- `elif`: additional condition;
- `else`: default action when previous conditions are false.


````python
# Example: classify a student mark.
mark = 68

if mark >= 70:
    classification = "Distinction"
elif mark >= 60:
    classification = "Merit"
elif mark >= 50:
    classification = "Pass"
else:
    classification = "Fail"

print(classification)
````

### `match`

`match` was introduced in Python 3.10. It is useful when one value needs to be compared with several patterns.


````python
# Example: respond to a command.
command = "start"

match command:
    case "start":
        print("Starting...")
    case "stop":
        print("Stopping...")
    case "pause":
        print("Pausing...")
    case _:
        # The underscore means the default case.
        print("Unknown command")
````

## 5. `for` loops

A `for` loop repeats an action for every item in an iterable object, such as a list, tuple, string, dictionary, or range.


````python
# Loop through a list.
courses = ["Python", "R", "SQL"]

for course in courses:
    print(course)
````

````python
# range(5) gives 0, 1, 2, 3, 4.
for i in range(5):
    print(i)
````

### `enumerate()`

Use `enumerate()` when you need both the index and the value.


````python
# enumerate() returns index-value pairs.
courses = ["Python", "R", "SQL"]

for index, course in enumerate(courses):
    print(index, course)
````

### `zip()`

`zip()` is used when we want to loop over two or more sequences **at the same time**.

This is common when different lists contain related information. For example, one list may contain student names and another list may contain their marks. The first name belongs with the first mark, the second name belongs with the second mark, and so on.

The general pattern is:

```python
for item_from_first_list, item_from_second_list in zip(first_list, second_list):
    # use the paired values here
```

Important point: `zip()` stops when the shortest sequence ends. If one list has more items than the other, the extra items are ignored.


````python
# Example: use zip() to loop through two related lists.
# Each name is paired with the mark in the same position.

students = ["Sara", "Omar", "Mina"]
marks = [72, 58, 81]

# zip(students, marks) creates pairs:
# ("Sara", 72), ("Omar", 58), ("Mina", 81)
for student, mark in zip(students, marks):
    print(student, "scored", mark)
````

#### Combining `zip()` with conditions

Inside the loop, the paired values can be used in the same way as any other loop variables. For example, we can classify each student's mark while looping through the two lists together.


````python
# Example: combine zip() with an if-elif-else statement.
# This allows us to process paired information in one loop.

students = ["Sara", "Omar", "Mina", "Leo"]
marks = [72, 58, 81, 46]

for student, mark in zip(students, marks):
    # Decide the classification for the current mark.
    if mark >= 70:
        classification = "Distinction"
    elif mark >= 60:
        classification = "Merit"
    elif mark >= 50:
        classification = "Pass"
    else:
        classification = "Fail"

    # Print one clear sentence for the current student.
    print(f"{student}: {mark} — {classification}")
````

#### Using `zip()` with three lists

`zip()` can combine more than two sequences. The number of loop variables must match the number of sequences passed to `zip()`.


````python
# Example: loop over three related lists at the same time.
# Here, each row is made from one item from each list.

students = ["Sara", "Omar", "Mina"]
courses = ["Python", "R", "SQL"]
marks = [72, 58, 81]

for student, course, mark in zip(students, courses, marks):
    print(f"{student} studied {course} and scored {mark}.")
````

#### `zip()` and unequal lengths

If the lists do not have the same length, `zip()` only returns complete pairs. This protects us from getting an index error, but it also means that extra values are silently ignored.


````python
# Example: zip() stops at the shortest list.
# The last student is not printed because there is no matching mark.

students = ["Sara", "Omar", "Mina", "Leo"]
marks = [72, 58, 81]

for student, mark in zip(students, marks):
    print(student, mark)
````

## 6. `while` loops

A `while` loop continues while a condition remains true. Use it when the number of repetitions is not known in advance.


````python
# Count from 1 to 5.
count = 1

while count <= 5:
    print(count)
    count = count + 1  # Important: update the condition variable.
````

A common mistake is to write a `while` loop whose condition never becomes false. That creates an infinite loop.


## 7. `break`, `continue`, and loop `else`

- `break` stops the loop immediately.
- `continue` skips to the next iteration.
- `else` after a loop runs only if the loop finishes without `break`.


````python
# break example: stop when the target is found.
numbers = [2, 4, 6, 8, 10]
target = 6

for number in numbers:
    if number == target:
        print("Found:", number)
        break
````

````python
# continue example: skip odd numbers.
numbers = [1, 2, 3, 4, 5, 6]

for number in numbers:
    if number % 2 != 0:
        continue
    print(number)
````

````python
# loop else example: runs only if break is not used.
numbers = [1, 3, 5, 7]
target = 4

for number in numbers:
    if number == target:
        print("Found")
        break
else:
    print("Target was not found")
````

## 8. List comprehensions

A list comprehension is a compact way to create a list.

General form:

```python
new_list = [expression for item in iterable]
```

With a condition:

```python
new_list = [expression for item in iterable if condition]
```


````python
# Normal loop version.
squares = []

for number in range(1, 6):
    squares.append(number ** 2)

print(squares)
````

````python
# List comprehension version.
squares = [number ** 2 for number in range(1, 6)]

print(squares)
````

````python
# Keep only even numbers.
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
````

## 9. Functions

A function is a reusable block of code. Functions help avoid repetition and make programs easier to read, test, and debug.


````python
# Define a simple function.
def greet(name):
    """Return a greeting message for a name."""
    return f"Hello, {name}!"

# Call the function.
greet("Ali")
````

````python
# Function with two parameters.
def calculate_mean(a, b):
    """Return the mean of two numbers."""
    return (a + b) / 2

result = calculate_mean(10, 20)
print(result)
````

## 10. Functions are first-class citizens

In Python, functions are first-class objects. This means we can:

- assign a function to a variable;
- pass a function as an argument;
- return a function from another function;
- store functions in lists or dictionaries.


````python
# Assign a function to a variable.
def double(x):
    """Return x multiplied by 2."""
    return x * 2

operation = double

print(operation(5))
````

````python
# Pass a function as an argument.
def apply_operation(values, operation):
    """Apply operation to every item in values."""
    results = []
    for value in values:
        results.append(operation(value))
    return results

numbers = [1, 2, 3, 4]
print(apply_operation(numbers, double))
````

## 11. Inner functions

An inner function is defined inside another function. This is useful when a helper function is only needed locally.


````python
def summarise_marks(marks):
    """Return classifications for a list of marks."""

    def classify(mark):
        """Classify one mark."""
        if mark >= 70:
            return "Distinction"
        elif mark >= 60:
            return "Merit"
        elif mark >= 50:
            return "Pass"
        else:
            return "Fail"

    return [classify(mark) for mark in marks]

summarise_marks([45, 55, 65, 75])
````

## 12. Lambda expressions, `map`, `filter`, and `reduce`

A `lambda` expression creates a small anonymous function.

```python
lambda input: output
```

Use `lambda` only for short expressions. For longer logic, use `def`.


````python
# Lambda function.
square = lambda x: x ** 2

print(square(6))
````

````python
# map applies a function to every item.
numbers = [1, 2, 3, 4]
squared_numbers = list(map(lambda x: x ** 2, numbers))

print(squared_numbers)
````

````python
# filter keeps items that satisfy a condition.
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
````

````python
# reduce combines values into one result.
from functools import reduce

numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)

print(product)
````

## 13. Object-Oriented Programming \(OOP\)

Object-oriented programming organises code around objects.

An object can have:

- **attributes**: stored data;
- **methods**: functions attached to the object.

A **class** is a blueprint. An **object** is an instance of that class.


````python
# Define a class.
class Student:
    """Represent a student with a name and a mark."""

    def __init__(self, name, mark):
        # __init__ is the constructor.
        # It runs when a new Student object is created.
        self.name = name
        self.mark = mark

    def has_passed(self):
        # This method uses the object's stored mark.
        return self.mark >= 50

# Create an object.
student_1 = Student("Sara", 72)

print(student_1.name)
print(student_1.mark)
print(student_1.has_passed())
````

### `self`

Inside a class, `self` refers to the current object. For example, `self.name = name` stores the value of `name` inside the object.


## 14. Abstraction

Abstraction means hiding unnecessary internal details and exposing a clean interface.

For example, when we call `list.append()`, we do not need to know how Python internally resizes the list.


````python
class BankAccount:
    """A simple class showing abstraction."""

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        # Users call this method without needing internal details.
        if amount > 0:
            self.balance = self.balance + amount

    def withdraw(self, amount):
        # Only allow valid withdrawals.
        if 0 < amount <= self.balance:
            self.balance = self.balance - amount

account = BankAccount("Ali", 100)
account.deposit(50)
account.withdraw(30)

print(account.balance)
````

## 15. Inheritance

Inheritance allows a class to reuse and extend another class.

- Parent/base class: the general class.
- Child/subclass: the more specific class.


````python
class Person:
    """A general person."""

    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"My name is {self.name}."

class StudentPerson(Person):
    """A student is a specific type of person."""

    def __init__(self, name, course):
        # Reuse the parent constructor.
        super().__init__(name)
        self.course = course

    def introduce(self):
        # Override the parent method.
        return f"My name is {self.name}, and I study {self.course}."

student = StudentPerson("Mina", "Python")
print(student.introduce())
````

## 16. Polymorphism and duck typing

**Polymorphism** means that different objects can respond to the same method name in different ways.

**Duck typing** means Python focuses on whether an object has the required behaviour, rather than requiring a strict declared type.


````python
class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

# This function works for any object that has a speak() method.
def make_it_speak(animal):
    return animal.speak()

print(make_it_speak(Dog()))
print(make_it_speak(Cat()))
````

## 17. Dunder methods

Dunder methods have double underscores before and after their names. They are also called magic methods.

Common examples:

- `__init__`: constructor;
- `__str__`: controls `print(object)`;
- `__len__`: controls `len(object)`.


````python
class Course:
    """Represent a course with a title and a list of students."""

    def __init__(self, title, students):
        self.title = title
        self.students = students

    def __str__(self):
        # Controls the readable string representation.
        return f"Course: {self.title}"

    def __len__(self):
        # Controls len(course).
        return len(self.students)

course = Course("Introduction to Python", ["Sara", "Mina", "Omar"])

print(course)
print(len(course))
````

## 18. Exception handling

An exception is an error that occurs while the program is running.

Exception handling allows the program to respond to errors without immediately crashing.

Main keywords:

- `try`: code that might fail;
- `except`: what to do if a specific error occurs;
- `else`: what to do if no error occurs;
- `finally`: what to do regardless of success or failure.


````python
# Handle division by zero.
a = 10
b = 0

try:
    result = a / b
except ZeroDivisionError:
    print("Cannot divide by zero.")
````

````python
# Full try-except-else-finally structure.
value = "42"

try:
    number = int(value)
except ValueError:
    print("The value cannot be converted to an integer.")
else:
    print("Conversion succeeded:", number)
finally:
    print("This message always runs.")
````

Avoid catching errors too broadly with a bare `except:`. It can hide mistakes. Prefer specific exceptions such as `ValueError`, `TypeError`, `KeyError`, and `ZeroDivisionError`.


## 19. Context managers and the `with` statement

A **context manager** controls what happens when a block of code starts and finishes.

The most common use is working with files. A file must be opened before it can be used, and it should be closed after the program has finished using it.

The `with` statement handles this automatically.


### Why do we use `with`?

The basic structure is:

```python
with expression as variable:
    # code that uses variable
```

When Python enters the `with` block, it prepares the resource. When Python leaves the block, it cleans up the resource.

This is useful because cleanup happens even if an error occurs inside the block.


````python
# Example: writing text to a file using a context manager.

# The file is opened at the beginning of the with block.
# mode="w" means write mode: create the file or overwrite it if it already exists.
# encoding="utf-8" is a safe default for ordinary text files.
with open("example_context_manager.txt", mode="w", encoding="utf-8") as file:
    # Inside this block, the file is open and can be used.
    file.write("This file was created using a context manager.\n")
    file.write("The with statement will close the file automatically.\n")

# Outside the block, the file has already been closed.
# We do not need to call file.close() manually.
print("Finished writing the file.")
````

````python
# Example: reading the file back.

# mode="r" means read mode.
with open("example_context_manager.txt", mode="r", encoding="utf-8") as file:
    # read() returns the full content of the file as one string.
    content = file.read()

print(content)
````

### Is `with` necessary?

Technically, it is possible to open and close a file manually:

```python
file = open("example.txt", "r")
content = file.read()
file.close()
```

However, this is more error-prone. If an error happens before `file.close()`, the file may remain open.

For file handling, prefer `with open(...) as file:`. It is safer, cleaner, and standard Python practice.


````python
# Manual file handling is possible, but less safe.
# This example is shown for comparison only.

file = open("manual_file_example.txt", mode="w", encoding="utf-8")
file.write("This file is closed manually.\n")
file.close()  # We must remember to close the file ourselves.

print("Manual file handling finished.")
````

### Context managers beyond files

Context managers are not only for files. They are also used for resources that need setup and cleanup, such as:

- database connections;
- network connections;
- locks in parallel programming;
- temporary files or temporary folders;
- changing settings for a short block of code.

The key idea is always the same: **enter safely, do the work, exit cleanly**.


````python
# Example: using a temporary folder as a context manager.

# tempfile is a built-in Python module.
# TemporaryDirectory creates a folder that is automatically removed
# when the with block finishes.
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as folder_name:
    # Convert the folder name string into a Path object.
    folder = Path(folder_name)

    # Create a file inside the temporary folder.
    temporary_file = folder / "notes.txt"
    temporary_file.write_text("Temporary data", encoding="utf-8")

    # The file exists inside the with block.
    print("Temporary file exists inside block:", temporary_file.exists())

# After the block, Python has removed the temporary folder and its contents.
print("Temporary file exists after block:", temporary_file.exists())
````

### Common beginner mistakes

1. **Using the file outside the `with` block**  
   The file is closed after the block, so you should read or write inside the block.

2. **Forgetting indentation**  
   Code that uses the file must be indented under `with`.

3. **Using the wrong mode**  
   Common modes are:

   - `"r"`: read an existing file;
   - `"w"`: write a file from scratch;
   - `"a"`: append to the end of a file;
   - `"x"`: create a new file, but fail if it already exists.


````python
# Example: appending to an existing file.

# mode="a" means append mode.
# New content is added to the end of the file.
with open("example_context_manager.txt", mode="a", encoding="utf-8") as file:
    file.write("This line was appended later.\n")

# Read the updated file.
with open("example_context_manager.txt", mode="r", encoding="utf-8") as file:
    updated_content = file.read()

print(updated_content)
````

## 20. Logging

`print()` is useful for quick checks. For larger programs, `logging` is better because messages can be assigned severity levels.

Common levels:

- `DEBUG`: detailed diagnostic information;
- `INFO`: normal progress information;
- `WARNING`: unexpected but not fatal;
- `ERROR`: serious problem;
- `CRITICAL`: very serious problem.


````python
# Import Python's built-in logging module.
import logging

# Configure logging.
# force=True is useful in notebooks because logging may already be configured.
logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s",
    force=True
)

logging.info("The program has started.")
logging.warning("This is a warning message.")
logging.error("This is an error message.")
````

````python
# Use logging inside a function.
def safe_mean(values):
    """Return the mean of a non-empty list of numbers."""

    if len(values) == 0:
        logging.error("Cannot calculate the mean of an empty list.")
        return None

    logging.info("Calculating the mean.")
    return sum(values) / len(values)

print(safe_mean([10, 20, 30]))
print(safe_mean([]))
````

## Solution 7


````python
# Solution 7
def apply_to_list(values, operation):
    """Apply operation to each item in values."""
    results = []
    for value in values:
        results.append(operation(value))
    return results

def double(x):
    """Return x multiplied by 2."""
    return x * 2

numbers = [1, 2, 3, 4]
print(apply_to_list(numbers, double))
````

## Solution 9


````python
# Solution 9
class Rectangle:
    """Represent a rectangle with width and height."""

    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        """Return the area of the rectangle."""
        return self.width * self.height

    def perimeter(self):
        """Return the perimeter of the rectangle."""
        return 2 * (self.width + self.height)

rectangle = Rectangle(4, 7)

print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
````

````python
# Solution 10
def safe_divide(a, b):
    """Safely divide a by b."""
    try:
        return a / b
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

````python
# Solution 11
# Use zip() to pair each student with the mark in the same position.

students = ["Sara", "Omar", "Mina", "Leo"]
marks = [72, 58, 81, 46]

for student, mark in zip(students, marks):
    # Classify the current student's mark.
    if mark >= 70:
        classification = "Distinction"
    elif mark >= 60:
        classification = "Merit"
    elif mark >= 50:
        classification = "Pass"
    else:
        classification = "Fail"

    # Print a readable summary for this student.
    print(f"{student}: {mark} — {classification}")
````

## Optional challenge solution — `MarkBook`

This optional example combines dictionaries, classes, methods, conditions, and list comprehensions.


````python
class MarkBook:
    """Store student marks and provide summary methods."""

    def __init__(self):
        # Dictionary structure: name -> mark.
        self.marks = {}

    def add_student(self, name, mark):
        """Add or update one student's mark."""
        self.marks[name] = mark

    def average_mark(self):
        """Return the average mark, or None if there are no marks."""
        if len(self.marks) == 0:
            return None
        return sum(self.marks.values()) / len(self.marks)

    def passed_students(self):
        """Return names of students with marks of at least 50."""
        return [name for name, mark in self.marks.items() if mark >= 50]

    def __str__(self):
        """Return a readable summary."""
        return f"MarkBook with {len(self.marks)} students"

markbook = MarkBook()
markbook.add_student("Sara", 72)
markbook.add_student("Mina", 49)
markbook.add_student("Omar", 65)

print(markbook)
print("Average:", markbook.average_mark())
print("Passed:", markbook.passed_students())
````
